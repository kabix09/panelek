```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_wynagrodzenia = pd.read_csv('data/processed/wynagrodzenia.csv')
df_wolne_miejsca_pracy = pd.read_csv('data/processed/wolne_miejsca_pracy.csv')
df_szkolnictwo = pd.read_csv('data/processed/szkolnictwo.csv')
df_odsetek_produkcyjny = pd.read_csv('data/processed/odsetek_produkcyjny.csv')
df_ludnosci = pd.read_csv('data/processed/ludnosc.csv')
df_liczba_podmiotow_gospodarczych = pd.read_csv('data/processed/liczba_podmiotow_gospodarczych.csv')
df_inwestycje = pd.read_csv('data/processed/inwestycje.csv')
df_absolwenci = pd.read_csv('data/processed/absolwenci.csv')

df_meszkania = pd.read_excel('data/processed/power_query/mieszkania-oddane-do-użytkowania-na-10-tys.-ludności.xlsx', sheet_name='mieszkania oddane_ długi układ')
df_podmiot_z_udzialem_kap_zagranicznego = pd.read_excel('data/processed/power_query/tabela-układ-długi-danych.xlsx', sheet_name='podm. z udz. kap. zagraniczn.')
df_podmiot_nowo_zarejestr = pd.read_excel('data/processed/power_query/tabela-układ-długi-danych.xlsx', sheet_name='podm. nowo zarejestr.')
```

```python
df_meszkania = df_meszkania.rename(columns={'Rok': 'rok', 'Region': 'region', 'Mieszkania oddane do użytkowania na 10 tys. ludności': 'mieszkania_oddane_do_uzytkowania_na_10_tys_ludnosci'})
df_meszkania = df_meszkania.iloc[:, 1:]
display(df_meszkania.head())
```

```python
df_podmiot_z_udzialem_kap_zagranicznego = df_podmiot_z_udzialem_kap_zagranicznego.rename(columns={'Rok': 'rok', 'Jednosta terytorialna': 'region', 'podmioty z udziałem kapitału zagranicznego na 10 tys. mieszkańców': 'podmiot_z_udzialem_kap_zagranicznego_na_10_tys_mieszkancow'})
df_podmiot_z_udzialem_kap_zagranicznego = df_podmiot_z_udzialem_kap_zagranicznego.iloc[:, 1:]


display(df_podmiot_z_udzialem_kap_zagranicznego.head())
```

```python
df_podmiot_nowo_zarejestr = df_podmiot_nowo_zarejestr.rename(columns={'Rok': 'rok', 'podmioty nowo zarejestrowane na 10 tys. ludności w wieku produkcyjnym': 'podmiot_nowo_zarejestr_na_10_tys_ludnosci_w_wieku_produkcyjnym'})
df_podmiot_nowo_zarejestr = df_podmiot_nowo_zarejestr.iloc[:, 1:]

display(df_podmiot_nowo_zarejestr.head())
```

```python
from functools import reduce

dfs = [
    df_wynagrodzenia,
    df_wolne_miejsca_pracy,
    df_szkolnictwo,
    df_odsetek_produkcyjny,
    df_ludnosci,
    df_liczba_podmiotow_gospodarczych,
    df_inwestycje,
    df_absolwenci,
    df_podmiot_z_udzialem_kap_zagranicznego,
    df_meszkania,
    df_podmiot_nowo_zarejestr
]

df_merged = reduce(
    lambda left, right: pd.merge(left, right, on=['region', 'rok'], how='outer'),
    dfs
)
```

```python
df_merged.to_csv('data/processed/merged_data.csv', index=False)
```
