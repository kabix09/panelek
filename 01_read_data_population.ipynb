```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("data/raw/LUDN_1356_XTAB_20260424000018.xlsx", 
                   header=[0, 1, 2], 
                   sheet_name="TABLICA"
                   )
```

```python
display(df.head())
```

```python
# remove first column "numer"

df = df.iloc[:, 1:]
```

```python
df.columns = [
    "_".join([str(x) for x in col if pd.notna(x)])
    for col in df.columns
]

df.columns.values[0] = "region" # rename first column to "region"

df.columns = df.columns.str.replace(" ", "_")
```

```python
df = df.drop(index=0) # usuwanie wierwsza z "[osoba]"
```

```python
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
df_long = df.melt(
    id_vars="region",
    var_name="zmienna",
    value_name="wartosc"
)
display(df_long.head())
```

```python
df_long[["typ", "rok"]] = df_long["zmienna"].str.rsplit("_", n=1, expand=True)
df_long["rok"] = df_long["rok"].astype(int)

display(df_long.head())
```

```python
df_final = df_long.pivot(
    index=["region", "rok"],
    columns="typ",
    values="wartosc"
).reset_index()
```

```python
df_final = df_final.replace("-", np.nan)
```

```python
display(df_final)
```

```python
df_filtered = df_final.loc[:, df_final.columns.str.contains(r'^region$|^rok$|ogółem', regex=True)]
```

```python
display(df_filtered)
```

```python
df_filtered.to_csv("data/processed/ludnosc.csv", index=False)
```
