```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("data/raw/Odsetek ludności w wieku 15–64 lata z wykształceniem wyższym – różnica w stosunku do średniej krajowej (w _).xlsx", 
                   header=[0, 1, 2, 3], 
                   sheet_name="TABLICA"
                   )
```

```python
display(df.head())
```

```python
# remove first column "numer"

df = df.iloc[:, 1:]
```

```python
df.columns = [
    "_".join([str(x) for x in col if pd.notna(x)])
    for col in df.columns
]

df.columns.values[0] = "region" # rename first column to "region"

df.columns = df.columns.str.replace(" ", "_")
```

```python
df = df.drop(index=0) # usuwanie wierwsza z "[%]"
```

```python
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
import pandas as pd
import re

# 1. Załaduj dane (zakładając, że df to Twój DataFrame z Excela)
# df = pd.read_excel('twoj_plik.xlsx')

# 2. Unpivot (melt) - zamieniamy wszystkie kolumny (poza regionem) na dwa rzędy: klucz i wartość
df_long = df.melt(id_vars=['region'], var_name='full_key', value_name='wartosc')

# 3. Funkcja do wyciągania informacji z nazwy kolumny za pomocą Regex
# Wzór: [Wykształcenie]_[Grupa]_[Atrybut]_[Rok]
# Grupy: ogółem, mężczyźni, kobiety, miasto, wieś
# Atrybuty: wartość_liczbowa, wskaźnik_precyzji
pattern = r'^(.*)_(ogółem|mężczyźni|kobiety|miasto|wieś)_(wartość_liczbowa|wskaźnik_precyzji)_(\d{4})$'

def parse_metadata(col_name):
    match = re.match(pattern, col_name)
    if match:
        return match.groups()
    return (None, None, None, None)

# Stosujemy rozbicie nazw
metadata = df_long['full_key'].apply(parse_metadata)
df_long[['typ_wykształcenia', 'grupa', 'atrybut', 'rok']] = pd.DataFrame(metadata.tolist(), index=df_long.index)

# # Usuwamy wiersze, które nie pasowały do wzorca i kolumnę pomocniczą
# df_long = df_long.dropna(subset=['atrybut']).drop(columns=['full_key'])
```

```python
# 4. Pivot - przenosimy 'wartość_liczbowa' i 'wskaźnik_precyzji' do osobnych kolumn
df_final = df_long.pivot_table(
    index=['region', 'rok', 'typ_wykształcenia', 'grupa'],
    columns='atrybut',
    values='wartosc',
    aggfunc='first'
).reset_index()

# Czyszczenie nazw po operacji pivot
df_final.columns.name = None

# Opcjonalnie: zmiana typu roku na liczbę
df_final['rok'] = df_final['rok'].astype(int)
```

```python
# Wynik
display(df_final)
```

```python
import pandas as pd

# Zakładamy, że df_final to wynik z poprzedniego kodu, czyli:
# ['region', 'rok', 'typ_wykształcenia', 'grupa', 'wartość_liczbowa', 'wskaźnik_precyzji']

# 1. Tworzymy pomocniczą tabelę typu "długiego", łącząc typ wykształcenia i grupę
# Najpierw "rozpinamy" wskaźniki (wartość i precyzję) do formatu long
df_melted = df_final.melt(
    id_vars=['region', 'rok', 'typ_wykształcenia', 'grupa'],
    value_vars=['wartość_liczbowa', 'wskaźnik_precyzji'],
    var_name='rodzaj_miary',
    value_name='wartosc'
)

# 2. Tworzymy nową nazwę kolumny według Twojego wzoru: [wykształcenie]_[grupa]_[miara]
df_melted['new_col_name'] = (
    df_melted['typ_wykształcenia'] + "_" + 
    df_melted['grupa'] + "_" + 
    df_melted['rodzaj_miary']
)

# 3. Robimy Pivot, aby region i rok były wierszami, a nowe nazwy kolumnami
df_output = df_melted.pivot_table(
    index=['region', 'rok'],
    columns='new_col_name',
    values='wartosc',
    aggfunc='first'
).reset_index()

# 4. Czyszczenie kosmetyczne
df_output.columns.name = None

# Wynik
display(df_output)
```

```python
df_filtered = df_output.loc[:, df_output.columns.str.contains(r'^region$|^rok$|ogółem', regex=True)]
```

```python
display(df_filtered)
```

```python
df_filtered.to_csv("data/processed/odsetek_produkcyjny.csv", index=False)
```
