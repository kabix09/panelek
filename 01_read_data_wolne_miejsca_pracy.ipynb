```python
import pandas as pd
import numpy as np
import re
```

```python
df = pd.read_excel("data/raw/RYNE_2671_XTAB_20260425131033.xlsx", 
                   header=[0, 1], 
                   sheet_name="TABLICA"
                   )
```

```python
display(df.head())
```

```python
# remove first column "numer"

df = df.iloc[:, 1:]
```

```python
df.columns = [
    "_".join([str(x) for x in col if pd.notna(x)])
    for col in df.columns
]

df.columns.values[0] = "region" # rename first column to "region"

df.columns = df.columns.str.replace(" ", "_")
```

```python
df = df.drop(index=0) # usuwanie wierwsza z "[osoba]"
```

```python
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
df_long = df.melt(
    id_vars=['region'], 
    var_name='full_key', 
    value_name='wartosc'
)
```

```python
df_long = df_long.replace("-", np.nan)
```

```python
display(df_long)
```

```python
pattern = r'^(.*)_(\d{4})$'

def parse_cols(col_name):
    match = re.match(pattern, col_name)
    if match:
        return match.groups() # Zwraca (nazwa_zmiennej, rok)
    return (None, None)

# Stosujemy rozbicie
df_long[['zmienna', 'rok']] = df_long['full_key'].apply(
    lambda x: pd.Series(parse_cols(x))
)
```

```python
display(df_long)
```

```python
# 3. Pivot - robimy unikalne wiersze dla [region + rok]
# a zmienne 'ogółem' i 'bezrobotni...' trafiają do osobnych kolumn
df_final = df_long.pivot_table(
    index=['region', 'rok'],
    columns='zmienna',
    values='wartosc',
    aggfunc='first'
).reset_index()
```

```python
# Wynik
display(df_final)
```

```python
df_final.to_csv("data/processed/wolne_miejsca_pracy.csv", index=False)
```
